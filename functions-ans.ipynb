#### Explain the concept of generators in Python and how they are defined.
##### In Python, generators are a type of iterable that allow you to iterate over a sequence of values lazily, meaning they generate items one at a time and only when needed, rather than storing them all in memory at once. This makes generators especially useful for working with large datasets or infinite sequences.

```python
def count_up_to(n):
    count = 1
    while count <= n:
        yield count
        count += 1

# creating genetor obj
gen = count_up_to(5)
for number in gen:
    print(number)
```

#### What are the advantages of using generators over regular functions?
##### Advantages of Using Generators

* Memory Efficient: Do not store all values in memory.
* Lazy Evaluation: Values are produced only when needed.
* Infinite Sequences: Can represent infinite streams.

```python
def squares():
    n = 1
    while True:
        yield n * n
        n += 1

# Creating a generator object
gen = squares()

# Get the first 5 square numbers
for _ in range(5):
    print(next(gen))
```

#### What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
* `map()`- its purpose is to apply a function to all the elements and returns transformed iterables
* `filter()`-its purpose is to select elements meeting a purpose and returns filtered iterables
* `reduce()`-its pupose is to reduce to a single value via function and returns single result

```python
from functools import reduce

nums = [1, 2, 3, 4]

# map: square numbers
print(list(map(lambda x: x**2, nums)))  # [1, 4, 9, 16]

# filter: keep even numbers
print(list(filter(lambda x: x % 2 == 0, nums)))  # [2, 4]

# reduce: sum of all numbers
print(reduce(lambda x, y: x + y, nums))  # 10
```

```python
# 1)Write a Python function that takes a list of numbers as input and returns
# the sum of all even numbers in the list
def sum_of_even_numbers(numbers):
    sum_of_even_is = 0
    for num in numbers:
        if num % 2 == 0:
            sum_of_even_is += num
    return sum_of_even_is
```

```python
# 4)Write a Python function that checks if a given number is prime or not from 1 to 200.
def is_prime(n):
    '''So, a prime number is the number which is only divisible by 1 and itself
    2 special cases also noted
    * prime numbers are not divisible by 2
    * 2 is the only even number which is prime.'''
    if n <= 1:
        return False  # 1 and below are not prime
    if n == 2:
        return True  # 2 is the only even prime number
    if n % 2 == 0:
        return False  # All other even numbers are not prime
    for i in range(3, int(n ** 0.5) + 1, 2):
        if n % i == 0:
            return False
    return True

#running a loop from 1 to 200(n-1)
for num in range(1, 201):
    if is_prime(num):
        print(num, end=",")

```

```python
# 5)Create an iterator class in Python that generates the Fibonacci
# sequence up to a specified number of terms
class FibonacciIterator:
    def __init__(self, max_terms):
        self.max_terms = max_terms  # total number of terms to generate
        self.count = 0              # current term index
        self.a, self.b = 0, 1       # starting values of the Fibonacci sequence

    def __iter__(self):
        return self  # an iterator must return itself

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration  # stop the iteration after max_terms
        if self.count == 0:
            self.count += 1
            return 0
        elif self.count == 1:
            self.count += 1
            return 1
        else:
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return self.a
fib = FibonacciIterator(20)
output = ','.join(str(num) for num in fib)
print(output)
```

```python
# 6)Write a generator function in Python that yields the powers of 2 up to a given exponent.
def power_of_two(n):
    for i in range(n):
        yield 2 ** i
```

```python
# 7)Implement a generator function that reads a file line by line and yields each
# line as a string.
def read_file_line_by_line(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line.strip()
```

```python
# 9)Write a Python program that uses `map()` to convert a list of temperatures
# from Celsius to Fahrenheit

celsius = [0, 20, 37, 100]# List of temperatures in Celsius

# Convert to Fahrenheit using map()
fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))

print("Fahrenheit: ", fahrenheit)
```

```python
# 10)Create a Python program that uses `filter()` to remove all the vowels from a given string.
def rm_vowel_str(string):
    vowels = 'aeiouAEIOU' #these elements will be removed from the string
    return ''.join(filter(lambda x: x not in vowels, string))
```
