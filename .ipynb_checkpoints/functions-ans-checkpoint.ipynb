#### What is the difference between a function and a method in Python?
* Function: A block of reusable code, defined using def, that can be called independently.
* Method: A function that is associated with an object and called using the dot (.) operator.

####  Explain the concept of function arguments and parameters in Python.
##### Function Arguments and Parameters in Python
* Parameters are the variables listed in a function definition.
*  Arguments are the values passed to the function when it is called.

```python
#Example
def add(a, b):  # a and b are parameters
    return a + b

result = add(5, 3)  # 5 and 3 are arguments
```

#### What are the different ways to define and call a function in Python?
* One way is the use of `def` keyword.
* Other way is the use of `lambda` keyword.

```python
def saying_HEY():          # def keyword is used; function name is given
    print("Hello")

lmda_says_hello = lambda: print("Hello from lambda")  #lambda is used, anonymous function.
```

```python
#ways of calling the function
saying_HEY()
lmda_says_hello()
```

#### What is the purpose of the `return` statement in a Python function?
 Purpose of the return Statement in a Python Function

* It is used to send back a result from a function to the caller.

* Ends the function execution.

```python
def square(x):
    return x * x

print(square(4))  # Output: 16
```

#### What are iterators in Python and how do they differ from iterables?
Iterators vs Iterables

* Iterable: An object that can be looped over (e.g., list, tuple).

* Iterator: An object with a `__next__()` method that returns the next item.

```python
my_list = [1, 2, 3]     # Iterable
my_iter = iter(my_list) # Iterator

print(next(my_iter))  # Output: 1
```
